```python
# ==============================================================================
# CELDA 1: Simulador SOC (Flujo de Escalamiento + Alerta SIEM)
# ==============================================================================
import json
import ipywidgets as widgets
from IPython.display import display, Markdown

# --- A) Deslizador con las 5 etapas del flujo del SOC ---
def flujo_soc_interactivo(paso):
    etapas = {
        1: {"Fase": "1. Recepción y Registro de Alerta", "Rol": "Analista SOC Nivel 1 (N1)",
            "Accion": "Recepción de alerta inusual desde SIEM/EDR y creación del ticket de incidente.",
            "Tecnologias": "SIEM, EDR, Sistema de Gestión de Tickets", "Estado": "🟡 Triaje Inicial"},
        2: {"Fase": "2. Triaje e Investigación Inicial (N1 -> N2)", "Rol": "Analista N1 / N2",
            "Accion": "Evaluación de gravedad. Si supera capacidad de N1, se escala formalmente a N2.",
            "Tecnologias": "Sistema de Tickets, SIEM", "Estado": "🟠 Escalado a N2"},
        3: {"Fase": "3. Investigación Detallada y Forense (N2 -> N3)", "Rol": "Analista N2 / N3",
            "Accion": "Análisis causa-raíz y mitigación. Si es complejo/desconocido, escala a N3.",
            "Tecnologias": "Herramientas Forenses (Memory/Disk Analysis)", "Estado": "🔴 Escalado a N3"},
        4: {"Fase": "4. Investigación Avanzada e Ingeniería Inversa", "Rol": "Analista SOC Nivel 3 (N3)",
            "Accion": "Análisis avanzado de malware, reversing e identificación de IoCs.",
            "Tecnologias": "Sandbox, Desensambladores, Malware Analysis Tools", "Estado": "🔬 Análisis Profundo"},
        5: {"Fase": "5. Resolución, Documentación y Lecciones Aprendidas", "Rol": "Todos + Gerente del SOC",
            "Accion": "Cierre de ticket, reporte ejecutivo, actualización de reglas SIEM y mejoras.",
            "Tecnologias": "Sistema de Tickets, Plataforma de Documentación / KB", "Estado": "🟢 Cierre"}
    }
    info = etapas[paso]
    print("="*80)
    print(f"📌 ETAPA DEL SOC: {info['Fase']}")
    print(f"👤 Rol Principal:    {info['Rol']}")
    print(f"⚙️ Acción Clave:      {info['Accion']}")
    print(f"🛠️ Tecnologías:      {info['Tecnologias']}")
    print(f"🏷️ Estado del Ticket: {info['Estado']}")
    print("="*80 + "\n")

display(Markdown("#### 🎚️ Simulador del Flujo de Escalamiento (mueve el deslizador del 1 al 5):"))
widgets.interact(flujo_soc_interactivo,
                 paso=widgets.IntSlider(min=1, max=5, step=1, value=1, description="Paso del SOC:"))

# --- B) Botón de Alerta SIEM (caso de estudio) ---
def generar_alerta(b):
    display(Markdown("### 🚨 ALERTA RECIBIDA EN EL SIEM (Ticket #INC-2024-089)"))
    alerta = {
        "timestamp": "2024-05-12T03:14:22Z",
        "severity": "HIGH",
        "source_ip": "192.168.10.45 (WS-RRHH-04)",
        "destination_ip": "104.21.45.12 (External / Tor Exit Node)",
        "event_description": "Multiple failed login attempts followed by successful RDP connection and execution of encoded PowerShell script.",
        "ioc_detected": "vssadmin.exe delete shadows /all /quiet",
        "user_account": "j.perez@empresa.local"
    }
    print(json.dumps(alerta, indent=4, ensure_ascii=False))
    display(Markdown("**Contexto:** El SIEM correlacionó eventos del EDR y el Firewall perimetral. RRHH reporta lentitud y archivos con extensiones inusuales."))

display(Markdown("#### 📥 Presiona el botón para recibir la alerta del caso:"))
boton = widgets.Button(description="📥 Recibir Alerta del SIEM", button_style='danger', icon='exclamation-triangle')
boton.on_click(generar_alerta)
display(boton)
```
