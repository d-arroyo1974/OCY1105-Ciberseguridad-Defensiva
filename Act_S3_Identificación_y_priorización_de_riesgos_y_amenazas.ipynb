```python
# ==============================================================================
# CELDA 1: Simulador Interactivo de Evaluación y Matriz de Riesgo (3x3)
# ==============================================================================
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
from ipywidgets import interact, widgets

def evaluador_riesgo_interactivo(probabilidad, impacto):
    # Matriz 3x3 de Criticidad
    # 1: Baja, 2: Media, 3: Alta
    p_map = {"Baja": 1, "Media": 2, "Alta": 3}
    i_map = {"Bajo": 1, "Medio": 2, "Alto": 3}

    p_val = p_map[probabilidad]
    i_val = i_map[impacto]
    score = p_val * i_val

    if score >= 6:
        nivel = "🔴 CRÍTICO / ALTO"
        color_bg = "red"
    elif score >= 3:
        nivel = "🟠 MEDIO"
        color_bg = "orange"
    else:
        nivel = "🟢 BAJO"
        color_bg = "green"

    print("="*75)
    print(f"📊 EVALUACIÓN DE RIESGO: Probabilidad [{probabilidad}] x Impacto [{impacto}]")
    print(f"🎯 NIVEL DE RIESGO CALCULADO: {nivel} (Puntuación: {score}/9)")
    print("="*75 + "\n")

    # Dibujar Matriz 3x3
    matrix_data = [[1, 2, 3], [2, 4, 6], [3, 6, 9]]
    df_matrix = pd.DataFrame(matrix_data, index=["P. Baja", "P. Media", "P. Alta"], columns=["I. Bajo", "I. Medio", "I. Alto"])

    plt.figure(figsize=(5, 3.5))
    sns.heatmap(df_matrix, annot=True, cmap="YlOrRd", cbar=False, fmt="d", linewidths=1)
    plt.title("Matriz de Riesgo 3x3 (Probabilidad vs Impacto)")
    plt.show()

print("✅ Simulador de Matriz de Riesgo cargado correctamente:")
interact(
    evaluador_riesgo_interactivo,
    probabilidad=widgets.Dropdown(options=["Baja", "Media", "Alta"], value="Alta", description="Probabilidad:"),
    impacto=widgets.Dropdown(options=["Bajo", "Medio", "Alto"], value="Alto", description="Impacto:")
);
```
