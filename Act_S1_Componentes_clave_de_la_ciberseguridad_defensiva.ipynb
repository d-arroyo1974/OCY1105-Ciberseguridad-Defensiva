```python
# ==============================================================================
# CELDA 1: Inicialización de Librerías y Consola de Monitoreo SOC (PyTorch + Widgets)
# ==============================================================================
import torch
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
from ipywidgets import interact, widgets

def panel_interactivo_soc(caso, paso_linea_de_tiempo):
    """
    Simula la consola de monitoreo del SOC procesando eventos y tráfico mediante tensores.
    """
    if caso == "Caso 1: DDoS (Tech Solutions SPA)":
        horas = ["09:00 AM", "10:00 AM", "11:00 AM", "12:00 PM", "02:00 PM", "04:00 PM", "05:00 PM"]

        # Tensor de PyTorch que representa el tráfico entrante (peticiones / segundo)
        trafico_tensor = torch.tensor([150.0, 45000.0, 120000.0, 125000.0, 118000.0, 800.0, 160.0])
        estado_servicio = ["Óptimo", "Lento (Alerta)", "Caído (Colapso)", "Caído (Colapso)", "Caído (Colapso)", "Restaurando", "Normal"]

        idx = min(paso_linea_de_tiempo - 1, len(horas)-1)
        trafico_normalizado = trafico_tensor / torch.max(trafico_tensor)

        print("="*75)
        print(f"🔥 [SOC Dashboard - PyTorch Backend] Estado a las {horas[idx]}: {estado_servicio[idx]}")
        print(f"📈 Volumen en Tensor: {trafico_tensor[idx].item():.0f} req/sec")
        print(f"⚡ Carga Operativa Normalizada: {trafico_normalizado[idx].item():.2%}")
        print("="*75 + "\n")

        plt.figure(figsize=(9, 4))
        plt.plot(horas[:paso_linea_de_tiempo], trafico_tensor[:paso_linea_de_tiempo].numpy(), color="darkred", marker="o", linewidth=2.5)
        plt.axhline(y=10000, color='orange', linestyle='--', label='Umbral de Alerta SIEM')
        plt.xlim(horas[0], horas[-1])
        plt.ylim(0, 140000)
        plt.title(f"Línea de Tiempo del Ataque DDoS - Monitoreo en Vivo ({horas[idx]})", fontsize=12, fontweight='bold')
        plt.ylabel("Peticiones por Segundo (req/sec)")
        plt.xlabel("Hora del Día")
        plt.legend(loc="upper left")
        plt.grid(True, linestyle="--", alpha=0.6)
        plt.show()

    elif caso == "Caso 2: Ransomware (InnovateTech Inc.)":
        alertas_totales = [
            {"Hora": "08:50 AM", "Componente": "Endpoint (EDR)", "Alerta": "Descarga de binario sospechoso (.exe)", "Severidad": 2},
            {"Hora": "08:55 AM", "Componente": "Sistema Local", "Alerta": "Ejecución: vssadmin.exe delete shadows", "Severidad": 3},
            {"Hora": "09:00 AM", "Componente": "FileServer-01", "Alerta": "Modificación masiva de extensiones a (.locked)", "Severidad": 4},
            {"Hora": "09:05 AM", "Componente": "Red (Firewall)", "Alerta": "Tráfico saliente hacia Red TOR (C2 Server)", "Severidad": 4}
        ]

        alertas_visibles = alertas_totales[:paso_linea_de_tiempo]
        severidades_tensor = torch.tensor([a["Severidad"] for a in alertas_visibles])

        print("="*75)
        print(f"🚨 [SIEM Core Engine] Nivel Crítico Global (PyTorch Metric Tensor): {severidades_tensor.tolist()}")
        print("="*75)

        mapa_severidad = {2: ("🟡 Media", "yellow"), 3: ("🟠 Alta", "orange"), 4: ("🔴 CRÍTICA", "red")}
        for row in alertas_visibles:
            texto_sev, _ = mapa_severidad[row["Severidad"]]
            print(f"⏱️ [{row['Hora']}] | 💻 [{row['Componente']}] | ⚠️ {row['Alerta']} -> Status: {texto_sev}")

        plt.figure(figsize=(6, 2.5))
        df_alertas = pd.DataFrame(alertas_visibles)
        df_alertas['Severidad'].value_counts().plot(kind='barh', color=['red', 'orange', 'yellow'][:len(df_alertas)])
        plt.title("Alertas Activas por Severidad en SIEM")
        plt.xlabel("Cantidad de Incidentes")
        plt.xlim(0, 4)
        plt.grid(axis='x', linestyle='--', alpha=0.7)
        plt.show()

print("✅ Panel interactivo del SOC cargado correctamente. Utiliza los controles deslizantes a continuación:")
interact(
    panel_interactivo_soc,
    caso=widgets.Dropdown(
        options=["Caso 1: DDoS (Tech Solutions SPA)", "Caso 2: Ransomware (InnovateTech Inc.)"],
        description="Caso:"
    ),
    paso_linea_de_tiempo=widgets.IntSlider(min=1, max=7, step=1, value=1, description="Línea Tiempo:")
);
```
